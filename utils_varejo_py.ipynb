```python
# --- INSTALAÇÃO RÁPIDA ---
!pip install pandas plotly -q

# --- GERAÇÃO E ANÁLISE EM TEMPO REAL ---
import pandas as pd
import numpy as np
import plotly.express as px
from datetime import datetime, timedelta

print("⏳ Gerando dados e processando... Aguarde 5 segundos.")

# 1. Criar Dados Simulados (Motor Embutido)
np.random.seed(42)
n_rows = 10000
dates = [datetime(2024, 1, 1) + timedelta(days=int(i)) for i in np.random.randint(0, 365, n_rows)]

df = pd.DataFrame({
'data': dates,
'regiao': np.random.choice(['Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste'], n_rows, p=[0.1, 0.2, 0.4, 0.2, 0.1]),
'setor': np.random.choice(['Alimentos', 'Saude', 'Eletronicos', 'Vestuario'], n_rows),
'valor': np.random.uniform(20, 2000, n_rows),
'pagamento': np.random.choice(['Pix', 'Cartao', 'Dinheiro'], n_rows)
})

# Injetar Tendências Reais (A "Mágica")
# Saúde cresce no 1º semestre
df.loc[(df['setor'] == 'Saude') & (df['data'] < datetime(2024, 7, 1)), 'valor'] *= 1.15
# Páscoa no Norte (Março/Abril)
df.loc[(df['regiao'] == 'Norte') & (df['data'].dt.month.isin([3, 4])), 'valor'] *= 1.20

df['mes'] = df['data'].dt.month_name()
df['ticket'] = (df['valor'] / np.random.randint(1, 10, n_rows)).round(2)

# 2. Mostrar KPIs Rápidos
total = df['valor'].sum()
ticket = df['ticket'].mean()
print(f"✅ Pronto! {len(df)} transações analisadas.")
print(f"💰 Faturamento: R$ {total:,.2f} | 🎫 Ticket Médio: R$ {ticket:,.2f}")

# 3. Gerar Gráficos Interativos Automáticos
print("📊 Renderizando gráficos...")

# Gráfico 1: Evolução
fig1 = px.line(df.groupby(['mes', 'setor'])['valor'].sum().reset_index(),
x='mes', y='valor', color='setor', title='📈 Vendas por Mês e Setor', markers=True)
fig1.show()

# Gráfico 2: Regiões
fig2 = px.bar(df.groupby('regiao')['valor'].sum().sort_values(ascending=False),
x='regiao', y='valor', color='valor', title='🌍 Ranking de Regiões', color_continuous_scale='Blues')
fig2.show()

# Gráfico 3: Pagamentos
fig3 = px.pie(df, names='pagamento', values='valor', title='💳 Share de Pagamentos', hole=0.4)
fig3.show()

# Gráfico 4: Prova da Sazonalidade (Páscoa no Norte)
df_norte = df[df['regiao'] == 'Norte']
fig4 = px.bar(df_norte.groupby(df_norte['data'].dt.month)['valor'].sum(),
title='🐰 Pico de Páscoa no Norte (Março/Abril)', color_continuous_scale='Oranges')
fig4.show()

print("🚀 Análise concluída. .")
```
